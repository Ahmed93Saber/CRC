```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
classes = ['LGD', 'HGD', 'CRC', 'Others']
accuracies_cpls = [81.08, 70.0, 98.82, 87.06]
accuracies_cpls_std = [2.77, 5.48, 1.44, 3.64]

accuracies_no_cpls = [81.08, 60.0, 99.41, 87.45]
accuracies_no_cpls_std = [6.25, 10.49, 1.18, 5.63]


# Set up positions and bar width
x = np.arange(len(classes))
width = 0.35 # Adjust width based on the number of models

# Create plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot bars for each model, offset by the width
ax.bar(x - width/2, accuracies_cpls, width, yerr=accuracies_cpls_std, label='Proposed approach with CPLS', capsize=5, alpha=0.8, color='#4C72B0')
ax.bar(x + width/2,         accuracies_no_cpls, width, yerr=accuracies_no_cpls_std, label='Proposed approach without CPLS', capsize=5, alpha=0.8, color='#DD8452')


# Labels, title, and customization
ax.set_ylabel('Accuracy')
ax.set_xlabel('Classes')
ax.set_title('Model Accuracy by Class with and without CPLS')
ax.set_xticks(x)
ax.set_xticklabels(classes)
ax.set_ylim(0, 110)
ax.legend()

# Grid
ax.set_axisbelow(True)
ax.yaxis.grid(True, linestyle='--', alpha=0.7)

# Display
plt.style.use('default')
plt.tight_layout()
plt.savefig("plots/barcharts/CPLS_impact.pdf", facecolor='white', transparent=False)
plt.show()
```
